```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/Attrition - Attrition.csv")
```

```python
df.head(2)
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
df.columns
```

```python
df=df.dropn()
```

```python
x=df.drop(columns=['Attrition'])
y=df['Attrition']
```

```python
x.info()
```

```python
categorical_features=['BusinessTravel','Department','EducationField','Gender',
'JobRole','MaritalStatus','Over18','OverTime']
```

```python
x_encoded=pd.get_dummies(x,columns=categorical_features,drop_first=True)
```

```python
x_encoded=x_encoded.astype(int)
```

```python
# x_encoded
```

```python
lb=LabelEncoder()
```

```python
y_encoded =lb.fit_transform(y)
y_encoded
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_encoded,y_encoded,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
sc=StandardScaler()
```

```python
x_train_scaled=sc.fit_transform(x_train)
x_test_scaled=sc.transform(x_test)
```

```python
# now we will create architecture of our model
```

```python
model=tf.keras.models.Sequential([
    tf.keras.layers.Dense(units=128,activation='relu',
                          input_dim=x_train_scaled.shape[1]),  #input layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=64,activation='relu'),     #hidden layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=1,activation='sigmoid')     #output layer
])
```

```python
# compile the model
```

```python
loss_fn=tf.keras.losses.BinaryCrossentropy()
metrics=['accuracy']
```

```python
learning_rate=0.001
momentum=0.9
optimizer=tf.keras.optimizers.SGD(learning_rate=learning_rate,momentum=momentum
                                  ,nesterov=True)
```

```python
model.compile(optimizer=optimizer,loss=loss_fn, metrics=metrics)
```

```python
# Train the model
```

```python
model.fit(x_train_scaled,
         y_train,
         epochs=50,
         batch_size=8,
         validation_split=0.1)
```

```python
loss , accuracy =model.evaluate(x_test_scaled,y_test)
```

```python
print(f'Test Loss={loss:.4f},Test Accuracy={accuracy:.4f}')
```

```python
import numpy as np
import pandas as pd
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("/content/titanic - titanic.csv")
```

```python
df.head(2)
```

```python

df=df.drop(columns=['Name','PassengerId','Ticket','Cabin'])
```

```python
df.isnull().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Fare']=df['Fare'].fillna(df['Fare'].mean())
```

```python
df.dtypes
```

```python
lb=LabelEncoder()
```

```python
df['Sex']=lb.fit_transform(df['Sex'])
df['Embarked']=lb.fit_transform(df['Embarked'])
```

```python
sc=StandardScaler()
```

```python
x=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
from imblearn.over_sampling import SMOTE

smote=SMOTE(sampling_strategy='minority')
x,y=smote.fit_resample(x,y)
y.value_counts()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train_scaled=sc.fit_transform(x_train)
x_test_scaled=sc.transform(x_test)
```

```python
model=tf.keras.models.Sequential([
    tf.keras.layers.Dense(units=128,activation='relu',
                          input_dim=x_train_scaled.shape[1]),  #input layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=64,activation='relu'),     #hidden layer
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=1,activation='sigmoid')     #output layer
])
```

```python
loss_fn=tf.keras.losses.BinaryCrossentropy()
metrics=['accuracy']
```

```python
learning_rate=0.001
momentum=0.9
optimizer=tf.keras.optimizers.SGD(learning_rate=learning_rate,momentum=momentum
                                  ,nesterov=True)
```

```python
model.compile(optimizer=optimizer,loss=loss_fn, metrics=metrics)
```

```python
model.fit(x_train_scaled,
         y_train,
         epochs=50,
         batch_size=8,
         validation_split=0.1)
```

```python
loss , accuracy =model.evaluate(x_test_scaled,y_test)
```

```python
print(f'Test Loss={loss:.4f},Test Accuracy={accuracy:.4f}')
```
