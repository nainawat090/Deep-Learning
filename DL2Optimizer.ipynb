```python
# TOTAL OPTIMIZERS==>
# in deep learning,optimizers are algorithm used to perameters of a neural network model during the training process.
# they play a crucial role in minimizing the loss function and improving the accuracy of the model.here are
# several popular optimizers along with their explainations:

# (1).stocastic gradiet descent(SGD):
# SGD is one of the simple optimizers. it updates the parametres in the direction of the negative
# graient of the loss function with respect to the current mini-batch of training data. however. sgd
# gas a limitation of converging slowly and can in local minima.

# (2).Momantam:
# the momontam optimizer builds upon SGD by adding a momontum term. it accumulates a fraction
# of the previous gradients to determine of the update . this helps to accelerate
# convergence and navigation through flat regions and local minima.

# (3).Nesterov Accelerated Gradient(NAG):
# NAG is an improvement over the momontum optimizer . it calculates the gradient not only based
# on the current parameters but also using an estimate of the future parameters . by looking ahead,
# NAG allows the optimizer to batter anticipate the momentums effect and adjust its update
# accordingly

# (4).Adarad:
# AdaGrad adapts the learning rate each parameters based on the historical gradients.it
# increases the learning rate for infrequent features and decreases it for frequent ones. this makes
# AdaGrad well-suited for sparse data but can cause the learning ate to become too small over time

# (5).RMSprop:
# RMSprop adresses the diminishing learning rate issue of AdaGrad by introducing an
# exponentially decaying average of squared gradients. By keeping a mooving average of the
# squared gradients. RMSprop normalizes the learning rates and improves convergence.

# Adam(Adaptive Moment Estimation):
# Adam combines the concepts of momentum and RMSprop. it maintains an exponentially decaying
# average of past gradients and squared gradients. Adam uses bias correction to account for the
# initialization bias in the first few iteration, making it perform well in practice across different deep
# learning tasks.
```
