```python
# Activation function---> activatin function ar mathematical functions applied to the outputs of individual neurons in a
# neural network. they introduce non-linearity into he network allowing it to learn and approximate
# complex relationships between inputs and outputs.
```

```python
# some commonly used activation function in deep learning:

# (1).Sigmoid function(logistic function):- it maps the input to a value between 0 and 1 . it was widely
# used in the past but is now less popular due to some drawbacks such as vanshing gradients.

# (2).Hyperbolic tangent functio(tanh):- similer to the sigmoid function, but it maps the input to a
# value between -1 and 1 . it is still used in some cases . but it also suffers from vanishing gradients.

# (3).Rectified Linear Unit(ReLU):- this function sets all negetive values to zero and keeps positive
# values unchanges . it is te most populr activation function in deep learning due to its simplicity
# and effectiveness in trainng deep neural networks.

# (4). Leaky ReLU:- This function is similer to ReLU but allows a small negative slope for negative
# input values . it mitigate the "dying ReLU" problem where some neurons can become
# peramenently inactive during traininng.

# (5).Parametric RLU(PReLU):PReLU:- PReLU is a generalization of ReLU tha introduce a learnable
# parameter to determie the slope of negative input values. it offers more flexibility and can
# improve model performance.

# (6).Exponential Linear Unit(ELU): ELU is a variation of ReLU that sllows negative values with a
# smooth exponential decay. it helps allevate the dying ReLU problem and can produce more
# robust models.

# (7).softmax:- Softmax is commonly used in the output layer of a neural network full multi-class
# classification problems. it normalizes the output values to represent probabilities, ensuring that the
# sum of all probabilities is 1.
```
