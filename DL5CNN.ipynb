```python
# Pooling Layer---> avoid overfitting from the dataset
# (1).Min Pooling layer---> it will return minimum number of pixel from array.
# (2). Max pooling Layer--- it will return maximum number of pixel from array.
# (3).GlobalAverage Pooling Layer---> it will return average of pixel from array.
```

```python
# (1).Min Pooling Layer--->
# [17,15,10,18]--->[10]
# [15,40,18,22]---->[15]
# [40,47,22,6]---->[6]
# [10,18,5,7]----->[5]
# [18,22,7,21]---->[7]
# [22,6,21,35]---->[6]

# it means when we apply min pooling operation on data then it will return this result
```

```python
# (2). Max pooling Layer--->

# [17,15,10,18]--->[18]
# [15,40,18,22]---->[40]
# [40,47,22,6]---->[47]
# [10,18,5,7]----->[18]
# [18,22,7,21]---->[22]
# [22,6,21,35]---->[35]

# it means when we apply max pooling operation on data then it will return this result
```

```python
# (3).GlobalAverage Pooling Layer--->

# [17,15,10,18]--->[15]
# [15,40,18,22]---->[24]
# [40,47,22,6]---->[47]
# [10,18,5,7]----->[10]
# [18,22,7,21]---->[17]
# [22,6,21,35]---->[21]
```
